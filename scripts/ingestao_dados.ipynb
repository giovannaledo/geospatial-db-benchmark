```python
import pandas as pd
from pymongo import MongoClient

print("Iniciando conexão...")
# Conectar ao banco
client = MongoClient('mongodb://localhost:27017/')
db = client['gis_benchmark']
collection = db['trajetorias']

# Ler as primeiras 5 linhas
df = pd.read_csv('data/SurveilDrone-Net23.csv').head(5)
print("Lendo dados com sucesso!")

# Inserir no MongoDB
collection.insert_many(df.to_dict('records'))
print("✅ Dados inseridos com sucesso no MongoDB!")
```
